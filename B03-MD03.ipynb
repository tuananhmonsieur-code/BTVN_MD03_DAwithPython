```python
"""
Bai 1
Viết hàm multiply tính tích các tham số truyền vào. Số lượng tham số truyền vào là tùy ý. Minh họa một số lời gọi hàm vừa viết.

Ví dụ:

multiply(1, 2, 3, 4) ? 24
multiply(12, 20, 0) ? 0
multiply(4, 5) ? 20

Lưu ý: hàm cần xử lý tối ưu. Nếu đang nhân các tham số để trả về kết quả và gặp số 0, cần trả về 0 luôn mà không tiến hành nhân tiếp.
"""
def multiply(*args):
    result = 1
    for arg in args:
        result *= arg
    return result

print(multiply(1, 2, 3, 4))
print(multiply(12,20,0))
print(multiply(4,5))
```

```python
"""
Bai 2
Viết hàm convert đổi từ độ F (Fahrenheit) sang độ C (Celcius). Minh họa sử dụng hàm vừa viết
"""
def convert(f):
    Celcius = (f-32)*5/9
    return round(Celcius,2)
print(convert(100))
```

```python
"""
Bai 3
Viết hàm tính lương compute_salary, với 2 tham số đầu vào là rate – số tiền trả cho mỗi giờ tiêu chuẩn và hours – số giờ làm việc.

Tham số rate là tùy chọn. Nếu không truyền vào, sẽ có giá trị mặc định là 1,000,000 (một triệu)

Với 40 giờ đầu tiên, mỗi giờ được trả chi phí tiêu chuẩn là rate. Từ những giờ làm việc OT tiếp theo, mỗi giờ cần trả chi phí là 1.5 rate

Minh họa ít nhất 3 lời gọi hàm vừa viết

Truyền vào đầy đủ các tham số như thông thường
Sử dụng tham số mặc định
Sử dụng tham số từ khóa
"""
def compute_salary(hours, rate=1000000):
    standard_hours = 40
    if hours <= standard_hours:
        salary = hours * rate
    else:
        overtime_hours = hours - standard_hours
        salary = (standard_hours * rate) + (overtime_hours * 1.5 * rate)
    return salary

# Minh họa ít nhất 3 lời gọi hàm vừa viết

# 1. Truyền vào đầy đủ các tham số như thông thường
print(f"Salary for 45 hours at 500,000/hour: {compute_salary(45, 500000):,.0f} VND")

# 2. Sử dụng tham số mặc định (rate)
print(f"Salary for 50 hours with default rate: {compute_salary(50):,.0f} VND")

# 3. Sử dụng tham số từ khóa
print(f"Salary for 35 hours using keyword arguments: {compute_salary(hours=35, rate=750000):,.0f} VND")
```

```python
"""
Bai4
Viết hàm calculation tính và trả về đồng thời cả tổng và hiệu hai số. Minh họa một số lời gọi hàm vừa viết.



Lưu ý: cần làm theo 2 cách: dùng và không dùng hàm lambda
"""

# Cách 1: Sử dụng hàm thông thường
def calculation_normal(num1, num2):
    sum_result = num1 + num2
    diff_result = num1 - num2
    return sum_result, diff_result

# Minh họa lời gọi hàm thông thường
a, b = calculation_normal(10, 5)
print(f"Với 10 và 5 (hàm thông thường): Tổng = {a}, Hiệu = {b}")

c, d = calculation_normal(20, 7)
print(f"Với 20 và 7 (hàm thông thường): Tổng = {c}, Hiệu = {d}")

# Cách 2: Sử dụng hàm lambda
calculation_lambda = lambda num1, num2: (num1 + num2, num1 - num2)

# Minh họa lời gọi hàm lambda
e, f = calculation_lambda(15, 3)
print(f"Với 15 và 3 (hàm lambda): Tổng = {e}, Hiệu = {f}")

g, h = calculation_lambda(8, 12)
print(f"Với 8 và 12 (hàm lambda): Tổng = {g}, Hiệu = {h}")
```

```python
"""
Bai4 Cach 3
"""
def calculate_product_quotient(num1, num2):
    product = num1 * num2
    if num2 == 0:
        quotient = "Cannot divide by zero"
    else:
        quotient = num1 / num2
    return product, quotient

# Minh họa lời gọi hàm
print("--- Minh họa hàm tính tích và thương ---")

# Ví dụ 1: Chia số dương
p1, q1 = calculate_product_quotient(10, 2)
print(f"Số thứ nhất: 10, Số thứ hai: 2 => Tích = {p1}, Thương = {q1}")

# Ví dụ 2: Chia số âm
p2, q2 = calculate_product_quotient(15, -3)
print(f"Số thứ nhất: 15, Số thứ hai: -3 => Tích = {p2}, Thương = {q2}")

# Ví dụ 3: Chia cho 0
p3, q3 = calculate_product_quotient(7, 0)
print(f"Số thứ nhất: 7, Số thứ hai: 0 => Tích = {p3}, Thương = {q3}")

# Ví dụ 4: Với số thập phân
p4, q4 = calculate_product_quotient(5.5, 2.5)
print(f"Số thứ nhất: 5.5, Số thứ hai: 2.5 => Tích = {p4}, Thương = {q4}")
```

```python
"""
Bai 5
Viết hàm say_hello(), tham số String, in ra lời chào bằng tiếng Việt/Anh/Đức/Ý (Xin chào, Hello, Hallo, Ciao) tùy theo đầu vào (“vn”, “en”, “ge”, “it”). Mặc định là chào tiếng Việt

Minh họa một số lời gọi hàm vừa viết.
"""

def say_hello(lang="vn"):
    greetings = {
        "vn": "Xin chào",
        "en": "Hello",
        "ge": "Hallo",
        "it": "Ciao"
    }
    # Lấy lời chào tương ứng, nếu không tìm thấy ngôn ngữ, mặc định là tiếng Việt
    greeting_message = greetings.get(lang.lower(), greetings["vn"])
    print(greeting_message)

# Minh họa một số lời gọi hàm vừa viết
print("--- Minh họa hàm say_hello ---")

say_hello()        # Mặc định tiếng Việt
say_hello("vn")    # Tiếng Việt
say_hello("en")    # Tiếng Anh
say_hello("ge")    # Tiếng Đức
say_hello("it")    # Tiếng Ý
say_hello("fr")    # Ngôn ngữ không xác định, mặc định tiếng Việt
```

```python
"""
Bai 6
Viết hàm age_filter: nhận tham số là ages - list hoặc tuple là các tuổi.Hàm cần trả về một list chỉ bao gồm các phần tử không nhỏ hơn 18 tuổi. Minh họa sử dụng hàm vừa viết



Cần viết  hàm age_filter theo 3 cách:

age_filter1 - Dùng vòng lặp for
age_filter2 - Dùng hàm filter của Python
age_filter3 - Dùng List comprehension
"""
# age_filter1 - Dùng vòng lặp for
def age_filter1(ages):
    filtered_ages = []
    for age in ages:
        if age >= 18:
            filtered_ages.append(age)
    return filtered_ages

# Minh họa sử dụng age_filter1
ages_data = [15, 22, 17, 30, 18, 12, 45]
print(f"Original ages (for loop): {ages_data}")
print(f"Filtered ages (for loop): {age_filter1(ages_data)}")

tuple_ages_data = (10, 19, 25, 16, 18)
print(f"Original ages (tuple, for loop): {tuple_ages_data}")
print(f"Filtered ages (tuple, for loop): {age_filter1(tuple_ages_data)}")
```

```python
# age_filter2 - Dùng hàm filter của Python
def age_filter2(ages):
    return list(filter(lambda age: age >= 18, ages))

# Minh họa sử dụng age_filter2
ages_data = [15, 22, 17, 30, 18, 12, 45]
print(f"Original ages (filter function): {ages_data}")
print(f"Filtered ages (filter function): {age_filter2(ages_data)}")

tuple_ages_data = (10, 19, 25, 16, 18)
print(f"Original ages (tuple, filter function): {tuple_ages_data}")
print(f"Filtered ages (tuple, filter function): {age_filter2(tuple_ages_data)}")
```

```python
# age_filter3 - Dùng List comprehension
def age_filter3(ages):
    return [age for age in ages if age >= 18]

# Minh họa sử dụng age_filter3
ages_data = [15, 22, 17, 30, 18, 12, 45]
print(f"Original ages (list comprehension): {ages_data}")
print(f"Filtered ages (list comprehension): {age_filter3(ages_data)}")

tuple_ages_data = (10, 19, 25, 16, 18)
print(f"Original ages (tuple, list comprehension): {tuple_ages_data}")
print(f"Filtered ages (tuple, list comprehension): {age_filter3(tuple_ages_data)}")
```

```python
"""
Bai 7
Viết hàm multi nhận 2 tham số đầu vào là (1) - một list hoặc tuple, và (2) - một số nguyên factor. Hàm này trả về một list, mỗi phần tử trong list này là tích của phần tử trong list/tuple ban đầu với factor.

Minh họa cách sử dụng hàm vừa viết
"""
def multi(input_collection, factor):
    result_list = []
    for item in input_collection:
        result_list.append(item * factor)
    return result_list

# Minh họa cách sử dụng hàm vừa viết
print("--- Minh họa hàm multi ---")

# Ví dụ với list
my_list = [1, 2, 3, 4, 5]
factor1 = 10
print(f"List gốc: {my_list}, Factor: {factor1} => Kết quả: {multi(my_list, factor1)}")

# Ví dụ với tuple
my_tuple = (10, 20, 30)
factor2 = 2
print(f"Tuple gốc: {my_tuple}, Factor: {factor2} => Kết quả: {multi(my_tuple, factor2)}")

# Ví dụ với số âm
another_list = [-1, 0, 5]
factor3 = -3
print(f"List gốc: {another_list}, Factor: {factor3} => Kết quả: {multi(another_list, factor3)}")

# Ví dụ với số thực
float_list = [1.5, 2.0, 3.5]
factor4 = 2.0
print(f"List gốc: {float_list}, Factor: {factor4} => Kết quả: {multi(float_list, factor4)}")
```

```python
"""
Bai 8
Viết hàm average trả về giá trị trung bình của các phần tử trong tham số đầu vào là một list/tuple
"""
def average(input_collection):
    total = sum(input_collection)
    return total / len(input_collection)
# Ví dụ với List:
my_list = [1, 2, 3, 4, 5]
print(f"List gốc: {my_list} => Giá trị trung bình: {average(my_list)}")

# Ví dụ với Tuple:
my_tuple = (10, 20, 30, 40, 50)
print(f"Tuple gốc: {my_tuple} => Giá trị trung bình: {average(my_tuple)}")
```

```python
"""
Bai 9
Viết hàm separate nhận một tham số là fullname - tên đầy đủ của người Việt, và trả về họ, tên đệm, và tên của người đó (1 tuple).

Tham số fullname truyền vào có thể là tên có 2, 3, 4, 5, … âm tiết. Ví dụ:

fullname = Trần Nam ? cần trả về (“Trần”, “”, “Nam”). Trường hợp này không có tên đệm nên trả về tên đệm là một xâu rỗng
fullname = Trần Hải Nam ? cần trả về (“Trần”, “Hải”, “Nam”)
fullname = Nguyễn Trần Hải Nam ? cần trả về (“Nguyễn”, “Trần Hải”, “Nam”)
fullname = Nguyễn Trần Hải Việt Nam ? cần trả về (“Nguyễn”, “Trần Hải Việt”, “Nam”)
"""
def separate(fullname):
  fullname = fullname.strip()
  names = fullname.split()
  if len(names) == 2:
    return (names[0], "", names[1])
  else:
    middle_name = " ".join(names[1:-1])
    return (names[0], middle_name, names[-1])
print("nguoi khong co ten dem:",separate("Trần Nam"))
print("nguoi co ten dem:",separate("Trần Hải Nam"))
print("nguoi co ten dem nhung 2 am tiet:",separate("Nguyễn Trần Hải Nam"))
print("nguoi co ten dem nhung co 3 am tiet:",separate("Nguyễn Trần Hải Việt Nam"))
```

```python
"""
Bai 10
Cho một list các xe hơi (car) có dạng như sau
"""
car_data = [
  {'car': 'Ford', 'year': 2005},
  {'car': 'Mitsubishi', 'year': 2000},
  {'car': 'BMW', 'year': 2019},
  {'car': 'VW', 'year': 2011}
]
"""
Cần viết hàm sort_car nhận tham số là list với cấu trúc như trên. Hàm này trả về list xe hơi đã sắp xếp xếp theo số năm sản xuất
"""
def sort_car(cars):
    return sorted(cars, key=lambda x: x['year'])

sorted_cars = sort_car(car_data)
print("Danh sách xe hơi đã sắp xếp theo năm sản xuất:")
for car in sorted_cars:
    print(car)
```

```python
"""
Bai 11
Cho một list các tham gia (participant) có dạng như sau
"""
participant_list = [

    {'name': 'Minh', 'age': 30, 'mark': 90},

    {'name': 'An', 'age': 28, 'mark': 70},

    {'name': 'Hoài', 'age': 25, 'mark': 90},

    {'name': 'Hải', 'age': 20, 'mark': 95},

]
"""
Cần viết hàm sort_participant nhận tham số là list với cấu trúc như trên. Hàm này trả về list tham gia đã sắp xếp xếp theo điểm và tuổi
"""
def sort_participant(participants):
    # Sắp xếp điểm cao đứng trước (-x['mark'] để giảm dần)
    # Nếu cùng điểm, ưu tiên người trẻ tuổi đứng trước (x['age'] để tăng dần)
    return sorted(participants, key=lambda x: (-x['mark'], x['age']))

sorted_participants = sort_participant(participant_list)
print("Danh sách tham gia đã sắp xếp theo điểm và tuổi:")
for participant in sorted_participants:
    print(participant)
```

```python
"""
Bai 12
Cho 1 list nhu sau:
"""
list1 = [

    {"name": "Nguyễn Hải Vân", "phone": "0976834523", "male": True},

    {"name": "Nguyễn Hải Phong", "phone": "0976834320", "male": True},

    {"name": "Nguyễn Thùy Lan", "phone": "0972834536", "male": False},

    {"name": "Trần Đức Vĩnh", "phone": "0979834543", "male": True},

    {"name": "Hà Lan Nhi", "phone": "0989824622", "male": False},

]
"""
Viết hàm sort_contact_list, nhận tham số là list với cấu trúc như trên. Hàm này cần trả về 1 list đã sắp xếp. Cách sắp xếp: sắp xếp theo tên trước, sau đó là theo họ, và sau cùng là tên đệm.
"""
def sort_contact_list(contacts):
    # Sắp xếp theo tên
    contacts.sort(key=lambda x: x['name'])
    # Sắp xếp theo họ
    contacts.sort(key=lambda x: x['name'].split()[0])
    # Sắp xếp theo tên đệm
    contacts.sort(key=lambda x: ' '.join(x['name'].split()[1:]))
    return contacts

sorted_contacts = sort_contact_list(list1)
print("Danh sách đã sắp xếp:")
for contact in sorted_contacts:
    print(contact)
```
