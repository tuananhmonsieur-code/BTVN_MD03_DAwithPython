# Lấy ngày giờ hiện tại

```python
import datetime

# Hiển thị thời gian hiện tại (múi giờ quốc tế)
# Thời gian bao gồm year, month, day, hour, minute, second, và microsecond.
x = datetime.datetime.now()
print(x)
```

# Timezone

```python
import pytz
zones = pytz.all_timezones

# Lấy ra các múi giờ
print(zones)

# Lấy ra giờ hiện tại theo múi giờ Việt Nam
vietnamTz = pytz.timezone("Asia/Saigon")
timeInVietNam = datetime.datetime.now(vietnamTz)
print("Việt Nam: ", timeInVietNam)


# Lấy ra giờ hiện tại theo múi giờ NewYork
newYorkTz = pytz.timezone("America/New_York")
timeInNewYork = datetime.datetime.now(newYorkTz)
print("Mỹ: ", timeInNewYork)


# Lấy ra giờ hiện tại theo múi giờ UTC
utcTz = pytz.timezone("UTC")
timeUTC = datetime.datetime.now(utcTz)
print("UTC: ", timeUTC)
```

# Tạo đối tượng datetime

```python
import datetime

# Tạo đối tượng datetime cơ bản: năm, tháng, ngày
x1 = datetime.datetime(2020, 5, 17)
print(x1)

# Tạo đối tượng datetime với thêm giá trị về giờ, phút, giây, micro giây
x2 = datetime.datetime(2020, 5, 17, 13, 59, 59, 5)
print(x2)


# Tạo đối tượng datetime với timezone Việt Nam
x3 = datetime.datetime(2020, 5, 17, 13, 59, 59, 5, pytz.timezone("Asia/Saigon"))
print(x3)


# Tạo đối tượng datetime với timezone quốc tế
x4 = datetime.datetime(2020, 5, 17, 13, 59, 59, 5, pytz.timezone("UTC"))
print(x4)
```

# Lấy ra các thông tin của datetime

```python
x = datetime.datetime.now(pytz.timezone("Asia/Saigon"))
print(x)


print(x.year, x.month, x.day, x.hour, x.minute, x.second, x.microsecond)
print(x.weekday()) # Thứ 2: 0, Thứ 3: 1, Thứ 4: 2, Thứ 5: 3, Thứ 6: 4, Thứ 7: 5, CN: 6

# Time zone
print(x.tzname())


# Đổi sang timezone khác
y = x.astimezone(pytz.timezone("UTC"))
print(y)
```

# Tạo xâu (String) định dạng từ đối tượng datetime

https://www.w3schools.com/python/python_datetime.asp

%a	Weekday, short version	Wed

%A	Weekday, full version	Wednesday

%w	Weekday as a number 0-6, 0 is Sunday	3

%d	Day of month 01-31	31

%b	Month name, short version	Dec

%B	Month name, full version	December

%m	Month as a number 01-12	12

%y	Year, short version, without century	18

%Y	Year, full version	2018

%H	Hour 00-23	17

%I	Hour 00-12	05

%p	AM/PM	PM

%M	Minute 00-59	41

%S	Second 00-59	08

%f	Microsecond 000000-999999	548513

%z	UTC offset	+0100

%Z	Timezone	CST

%j	Day number of year 001-366	365

%U	Week number of year, Sunday as the first day of week, 00-53	52

%W	Week number of year, Monday as the first day of week, 00-53	52

%c	Local version of date and time	Mon Dec 31 17:41:00 2018

%C	Century	20

%x	Local version of date	12/31/18

%X	Local version of time	17:41:00

%%	A % character	%

%G	ISO 8601 year	2018

%u	ISO 8601 weekday (1-7)	1

%V	ISO 8601 weeknumber (01-53)	01

```python
x = datetime.datetime.now(pytz.timezone("Asia/Saigon"))
print(x)

print(x.strftime("%A, %B, %H:%M %d-%m-%Y"))

vietnamWeekday = {"Monday": "Thứ Hai", "Tuesday": "Thứ Ba", "Wednesday": "Thứ Tư", "Thursday": "Thứ Năm", "Friday": "Thứ Sáu", "Saturday": "Thứ Bảy", "Sunday": "Chủ Nhật", }
print(vietnamWeekday[x.strftime("%A")] + x.strftime(", %B, %H:%M %d-%m-%Y"))
```

# Tạo datetime từ string

```python
x = datetime.datetime.strptime("Saturday, 05:07 05-11-2022", "%A, %I:%M %d-%m-%Y")
print(x)

print(x.strftime("%A, %d/%m/%Y %I:%M"))
```
