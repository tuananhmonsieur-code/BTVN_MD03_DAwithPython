## Ghi giá trị lên cell, thiết lập định dạng dữ liệu (số, ngày tháng, ...)

Các dạng format: https://openpyxl.readthedocs.io/en/stable/_modules/openpyxl/styles/numbers.html

```python
# Thiết lập giá trị cho 1 cell
new_sheet_1['A4'] = 4

# Định dạng kiểu dữ liệu
new_sheet_1['A5'].number_format = "yyyy-mm-dd h:mm:ss"
new_sheet_1['A5'] = datetime.datetime(2010, 7, 21)

new_sheet_1['A6'].number_format = "yyyy-mm-dd"
new_sheet_1['A6'] = datetime.datetime(2010, 7, 21)

new_sheet_1['A7'].number_format = '#,##0.00'
new_sheet_1['A7'] = 1733243248.3423423432123

save()
```

```python
# Ghi nhiều giá trị
for x in range(4, 11):
  for y in range(4, 11):
    new_sheet_1.cell(row=x, column=y).value = f"({x}, {y})"

save()
```

#Style cho cell

Font

```python
from openpyxl.styles import PatternFill, GradientFill, Border, Side, Alignment, Protection, Font

font = Font(name='Calibri',
            size=11,
            bold=True,
            italic=False,
            vertAlign="subscript", # {'superscript', 'subscript', 'baseline'} Không xem được trên Google sheet, chỉ xem được trên Excel
            underline='none',
            strike=False,
            color='FF0000')

new_sheet_1['D4'].font = font

save()
```

Fill

```python
fill = PatternFill(fill_type="solid", fgColor="3eb1de")
new_sheet_1['D5'].fill = fill

new_sheet_1['A14'].fill = GradientFill('linear', stop = ('85E4F7','4617F1')) # Không xem được trên Google sheet, chỉ xem được trên Excel

save()
```

Alignment

```python
alignment=Alignment(horizontal='general', # {'center’, 'left’, 'general’, 'justify’, 'centerContinuous’, 'right’, 'fill’, 'distributed’}
                    vertical='bottom', # {'bottom’, 'center’, 'justify’, 'top’, 'distributed’}
                    text_rotation=0,
                    wrap_text=False,
                    shrink_to_fit=True, # Không xem được trên Google sheet, chỉ xem được trên Excel
                    indent=0)

new_sheet_1['D7'].value = "Hello from PyOpenXL\nHello from PyOpenXL"
new_sheet_1['D7'].alignment = alignment

save()
```

# Tạo chart

Tạo dữ liệu

```python
from datetime import date

rows = [
    ['Date', 'Product A', 'Product B', 'Product C'],
    [date(2015,9, 1), 40, 30, 25],
    [date(2015,9, 2), 40, 25, 30],
    [date(2015,9, 3), 50, 30, 45],
    [date(2015,9, 4), 30, 25, 40],
    [date(2015,9, 5), 25, 35, 30],
    [date(2015,9, 6), 20, 40, 35],
]

for row in rows:
    new_sheet_2.append(row)

save()
```

Tạo line chart có trục hoành

```python
from openpyxl.chart.axis import DateAxis
from openpyxl.chart import (
   LineChart,
   Reference,
)

# Chart with date axis
c1 = LineChart()
c1.title = "Date Axis"
c1.style = 48
c1.y_axis.title = "Size"
c1.x_axis.title = "Date"

# Lấy dữ liệu từ ô (1, 2) đến ô (7, 4)
data = Reference(new_sheet_2, min_row=1, min_col=2, max_row=7, max_col=4)
c1.add_data(data, titles_from_data=True)

# Lấy dữ liệu từ ô (2, 1) đến ô (7, 1) làm các giá trị trên trục hoành
dates = Reference(new_sheet_2, min_row=2, min_col=1, max_row=7, max_col=1)
c1.set_categories(dates) # Thiết lập dữ liệu lấy được làm trục hoành

# Trình bày độ chia trên trục hoành
c1.x_axis.number_format = 'd-mmm'
c1.x_axis.majorTimeUnit = "days"

new_sheet_2.add_chart(c1, "A10")

save()
```

Tạo bar chart

```python
from openpyxl.chart import BarChart, Series, Reference

c2 = BarChart()
c2.title = "Date Axis"
c2.style = 2
c2.y_axis.title = "Size"
c2.x_axis.title = "Date"

c2.x_axis.number_format = 'd-mmm'
c2.x_axis.majorTimeUnit = "days"

# Lấy dữ liệu từ ô (1, 2) đến ô (7, 4)
data = Reference(new_sheet_2, min_row=1, min_col=2, max_row=7, max_col=4)
c2.add_data(data, titles_from_data=True)

dates = Reference(new_sheet_2, min_row=2, min_col=1, max_row=7, max_col=1)
c2.set_categories(dates)

new_sheet_2.add_chart(c2, "A27")

save()
```
