```python
"""
Bài 1
Tạo 1 biến. Tự tạo các ví dụ sử dụng các lệnh +=, -=, /=, %=, //=, *=, **=
Mỗi lệnh ít nhất một ví dụ.
"""
import math
var1 = 5
var2 = 10
var1 += var2
print(var1)
print(var2)
#Tự tạo các ví dụ sử dụng các lệnh -=,
var2 -= var1
print(var2)
#Tự tạo các ví dụ sử dụng các lệnh /=,
var2 /= var1
print(var2)
#Tự tạo các ví dụ sử dụng các lệnh %=,
var2 %= var1
print(var2)
#Tự tạo các ví dụ sử dụng các lệnh //=,
var2 //= var1
print(var2)
#Tự tạo các ví dụ sử dụng các lệnh *=,
var2 *= var1
print(var2)
#Tự tạo các ví dụ sử dụng các lệnh **=
var2 **=var1
print(var2)
```

```python
"""
Bai 3
Thực hành với hàm range

Cú pháp: range(start, stop, step), trả về 1 đối tượng range – chuỗi các số
start: Tùy chọn, là giá trị bắt đầu sinh trong chuỗi
stop: Giá trị sau giá trị cuối trong chuỗi được sinh
step: Tùy chọn, là khoảng giữa 2 số trong chuỗi.

Hàm range thường được dùng kết hợp với vòng lặp for

Ví dụ 1:
for i in range(5):
    print(i, end=" ")
print()

Kết quả in ra:
0 1 2 3 4
"""
#vi du
for i in range(5):
    print(i,end="")
print()
#In ra các số chia hết cho 5 từ 0 đến 100:
List1 = [*range(0,105,5)]
print(List1)

#Tạo 1 list bao gồm các số chia hết cho 3, ngược từ 99 về 0 theo 2 cách: chỉ dùng hàm range và dùng hàm reversed. In list đó ra màn hình.
List2 = [*range(0,100,3)]
print(list(reversed(List2)))
```

```python
"""
Bai 4
Với biến xâu name = “Nguyễn Hải Nam” lưu tên một người Việt nào đó. Giả thiết đó là tên đầy đủ tiêu chuẩn 3 chữ đơn với họ, tên đệm, tên. Sử dụng vòng lặp và phép so sánh xâu, tách họ, tên đệm và tên của người này, lưu vào 3 biến: firstName (tên), lastName (họ), và middleName (tên đệm)
"""
name = "Nguyễn Văn Nam"
lastName = ""
middleName = ""
firstName = ""

# Find the index of the first space
first_space_idx = -1
i = 0
while i < len(name):
    if name[i] == ' ':
        first_space_idx = i
        break
    i += 1

# Extract the last name (Họ)
lastName = name[:first_space_idx]

# Find the index of the second space, starting the search after the first space
second_space_idx = -1
j = first_space_idx + 1 # Start searching from the character after the first space
while j < len(name):
    if name[j] == ' ':
        second_space_idx = j
        break
    j += 1

# Extract the middle name (Tên đệm)
middleName = name[first_space_idx + 1 : second_space_idx]

# Extract the first name (Tên)
firstName = name[second_space_idx + 1 :]

print(f"Họ: {lastName}")
print(f"Tên đệm: {middleName}")
print(f"Tên: {firstName}")
```

```python
"""
Bai 6
Cho đầu vào là 1 list. Từ list này, cần tạo ra 2 list mới:
Một list chứa các phần tử có giá trị chẵn của list ban đầu
Một list chứa các phần tử có giá trị lẻ của list ban đầu
In ra 2 list đó

Dùng 2 cách: vòng lặp for thông thường, và dùng list comprehension

Ví dụ, với list1 = [10, 12, 23, 14, 51, 16, 72, 8, 90, 11], cần tạo ra 2 list con như sau
[10, 12, 14, 16, 72, 8, 90]
[23, 51, 11]
"""
list1 = [10, 12, 23, 14, 51, 16, 72, 8, 90, 11]
for i in range(len(list1)):
  if i %2 ==0:
    print("list cac phan tu co gia tri chan cua list ban dau:",list1)
  else:
    print("list cac phan tu co gia tri le cua list ban dau", list1)


```

```python
"""
Bai 7
Cho 1 list đầu vào là các số nguyên. Cần xử lý lọc list đó, tạo 1 list mới chỉ bao gồm các số thỏa mãn các điều kiện sau:
Chỉ lấy những số chia hết cho 5
Nếu gặp số lớn hơn 150, bỏ qua số đó, chuyển sang số tiếp theo
Nếu gặp số lớn hơn 500, kết thúc lọc
In ra màn hình list đầu ra

Ví dụ với list đầu vào là numbers = [12, 75, 150, 180, 145, 525, 50]
List đầu ra là filteredNumbers = [75, 150, 145]
"""
numbers = [12, 75, 150, 180, 145, 525, 50]
filteredNumbers = []
for num in numbers:
    if num > 500:
        break
    if num > 150:
        continue
    if num % 5 == 0:
      filteredNumbers += [num]
print(filteredNumbers)

```

```python
"""
Bài 8
Chỉ dùng vòng lặp for, in ra nội dung sau
1, 2, 3, 4, 5,
2, 4, 6, 8, 10,
3, 6, 9, 12, 15,
4, 8, 12, 16, 20,
5, 10, 15, 20, 25,
"""
for i in range(1, 6):
    for j in range(1, 6):
        # In kết quả trên cùng 1 hàng, sau mỗi số có thêm dấu phẩy và khoảng trắng
        print(i * j, end=", ")

    # Sau khi in xong 5 số của 1 hàng (hết vòng lặp j), mới xuống dòng
    print()

"""
Chỉ dùng vòng lặp while, vẫn in ra nội dung nói trên
"""
i = 1
while i <=5:
  j = 1
  while j<=5:
    print(i*j,end=', ')
    j+=1
  print()
  i+=1
```
