```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

try:
    # Thử đọc file cục bộ nếu đã tồn tại
    df_salary = pd.read_csv('/content/Salary_Data.csv')
    print("Đã đọc dữ liệu thành công từ tệp cục bộ.")
except FileNotFoundError:
    print("Không tìm thấy tệp cục bộ, tiến hành tạo dữ liệu mẫu và lưu thành /content/Salary_Data.csv...")
    # Tạo dữ liệu giả lập thực tế cho YearsExperience và Salary
    np.random.seed(42)
    years_exp = np.around(np.random.uniform(1.1, 10.5, 30), decimals=1)
    salary = np.around(30000 + years_exp * 9000 + np.random.normal(0, 5000, 30), decimals=0)

    df_salary = pd.DataFrame({
        'YearsExperience': years_exp,
        'Salary': salary
    })
    # Lưu lại cục bộ để sử dụng
    df_salary.to_csv('/content/Salary_Data.csv', index=False)

# Hiển thị cấu trúc dữ liệu
display(df_salary.head())

# Vẽ biểu đồ phân tán biểu diễn sự phụ thuộc của lương vào số năm kinh nghiệm
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_salary, x='YearsExperience', y='Salary', s=100, color='blue', edgecolor='black')
plt.title('Biểu đồ phân tán Lương theo Số năm kinh nghiệm')
plt.xlabel('Số năm kinh nghiệm')
plt.ylabel('Lương')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np

try:
    # Đọc dữ liệu từ file stocks.csv nếu tồn tại
    df_stocks = pd.read_csv('/content/stocks.csv')
    print("Đã đọc dữ liệu thành công từ tệp cục bộ.")
except FileNotFoundError:
    print("Không tìm thấy tệp stocks.csv cục bộ, đang tiến hành khởi tạo dữ liệu mẫu...")
    # Tạo dữ liệu giả lập cho 3 cổ phiếu: Google, Tesla, Amazon theo định dạng dọc (Long Format)
    dates = pd.date_range(start="2023-10-01", periods=10, freq="D").strftime('%Y-%m-%d').tolist() * 3
    symbols = ['Google'] * 10 + ['Tesla'] * 10 + ['Amazon'] * 10

    # Tạo giá mở cửa giả lập ngẫu nhiên
    np.random.seed(42)
    open_prices = np.concatenate([
        np.around(np.random.uniform(130, 140, 10), 2),  # Google
        np.around(np.random.uniform(240, 260, 10), 2),  # Tesla
        np.around(np.random.uniform(125, 135, 10), 2)   # Amazon
    ])

    df_stocks = pd.DataFrame({
        'date': dates,
        'symbol': symbols,
        'open': open_prices
    })

    # Lưu lại cục bộ dưới dạng stocks.csv
    df_stocks.to_csv('/content/stocks.csv', index=False)
    print("Đã khởi tạo và lưu file '/content/stocks.csv' thành công!")

# Hiển thị cấu trúc dữ liệu
print("\nCấu trúc dữ liệu ban đầu:")
display(df_stocks.head())
print("Các cột hiện có:", df_stocks.columns.tolist())
```

### Nhận xét cấu trúc dữ liệu
Dựa vào cấu trúc dữ liệu được hiển thị ở trên:
- Nếu mỗi dòng chứa các thông tin của nhiều cổ phiếu cùng một lúc trên các cột riêng biệt (ví dụ cột giá riêng cho Google, Tesla, Amazon), dữ liệu đang ở **dạng ngang (wide format)**.
- Nếu dữ liệu có một cột định danh tên cổ phiếu (như `symbol` hoặc `Name`) và cột giá chung (như `open`), dữ liệu đang ở **dạng dọc (long format)**.

Dưới đây là mã nguồn vẽ biểu đồ theo 2 phương pháp tương ứng sau khi xác định cấu trúc.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Cách 1: Giữ nguyên dữ liệu ban đầu để vẽ (Dữ liệu dạng dọc gốc)
plt.figure(figsize=(12, 6))
if 'symbol' in df_stocks.columns or 'Name' in df_stocks.columns:
    stock_col = 'symbol' if 'symbol' in df_stocks.columns else 'Name'
    sns.lineplot(data=df_stocks, x='date', y='open', hue=stock_col, marker='o')
    plt.title('Cách 1: Vẽ biểu đồ từ dữ liệu dạng dọc gốc (Long Format)')
else:
    open_cols = [col for col in df_stocks.columns if 'open' in col.lower() or any(s in col.lower() for s in ['goog', 'tsla', 'amzn'])]
    for col in open_cols:
        sns.lineplot(data=df_stocks, x='date', y=col, label=col, marker='o')
    plt.title('Cách 1: Vẽ biểu đồ trực tiếp từ dữ liệu dạng ngang gốc (Wide Format)')

plt.xlabel('Ngày')
plt.ylabel('Giá mở cửa')
plt.xticks(rotation=45)
plt.grid(True)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Cách 2: Chuyển đổi định dạng dữ liệu trước khi vẽ
plt.figure(figsize=(12, 6))

if 'symbol' in df_stocks.columns or 'Name' in df_stocks.columns:
    # Dữ liệu gốc là dạng dọc -> Chuyển sang dạng ngang (wide) bằng pivot rồi vẽ
    stock_col = 'symbol' if 'symbol' in df_stocks.columns else 'Name'
    df_wide = df_stocks.pivot(index='date', columns=stock_col, values='open').reset_index()

    # Vẽ từ dữ liệu đã chuyển đổi sang dạng ngang
    for stock in df_wide.columns:
        if stock != 'date':
            sns.lineplot(data=df_wide, x='date', y=stock, label=stock, marker='o')
    plt.title('Cách 2: Chuyển sang dạng ngang (Wide Format) rồi vẽ')
else:
    # Dữ liệu gốc là dạng ngang -> Chuyển sang dạng dọc (long) bằng melt rồi vẽ
    id_vars = ['date'] if 'date' in df_stocks.columns else [df_stocks.columns[0]]
    value_vars = [col for col in df_stocks.columns if col not in id_vars]

    df_long = pd.melt(df_stocks, id_vars=id_vars, value_vars=value_vars, var_name='Stock', value_name='Open_Price')

    sns.lineplot(data=df_long, x=id_vars[0], y='Open_Price', hue='Stock', marker='o')
    plt.title('Cách 2: Chuyển sang dạng dọc (Long Format) bằng melt rồi vẽ')

plt.xlabel('Ngày')
plt.ylabel('Giá mở cửa')
plt.xticks(rotation=45)
plt.grid(True)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Đọc dữ liệu từ file monthly_csv.csv
df_temp = pd.read_csv('/content/monthly_csv.csv')

print("Cấu trúc dữ liệu từ monthly_csv.csv:")
display(df_temp.head())
print("Các cột: ", df_temp.columns.tolist())
print("Thông tin dữ liệu:")
df_temp.info()
```

Dựa trên cấu trúc dữ liệu, có vẻ như cột `mean` đã chứa sự thay đổi nhiệt độ cho mỗi hàng. Tên cột `Source` cho biết tiêu chuẩn nào được sử dụng (GISTEMP hoặc GCAG). Để vẽ bản đồ nhiệt thể hiện sự ấm lên toàn cầu, chúng ta cần chuẩn bị dữ liệu sao cho năm là hàng, tháng là cột và giá trị nhiệt độ là màu sắc. Chúng ta sẽ tạo hai bản đồ nhiệt riêng biệt cho GISTEMP và GCAG để dễ so sánh.

Đầu tiên, chúng ta sẽ làm sạch và chuẩn bị dữ liệu:
- Chuyển đổi cột `Date` thành dạng datetime để trích xuất Năm và Tháng.
- Loại bỏ các hàng có giá trị `mean` bị thiếu.
- Chuyển đổi các giá trị tháng thành tên tháng hoặc số thứ tự phù hợp để hiển thị trên biểu đồ.

```python
# Chuẩn bị dữ liệu
# Chuyển đổi cột 'Date' thành datetime
df_temp['Date'] = pd.to_datetime(df_temp['Date'])

# Trích xuất năm và tháng
df_temp['Year'] = df_temp['Date'].dt.year
df_temp['Month'] = df_temp['Date'].dt.month

# Chuyển đổi số tháng thành tên tháng để hiển thị dễ đọc hơn
month_names = ['Tháng 1', 'Tháng 2', 'Tháng 3', 'Tháng 4', 'Tháng 5', 'Tháng 6',
               'Tháng 7', 'Tháng 8', 'Tháng 9', 'Tháng 10', 'Tháng 11', 'Tháng 12']
df_temp['Month_Name'] = df_temp['Month'].apply(lambda x: month_names[x-1])

# Loại bỏ các giá trị 'mean' bị thiếu nếu có
df_temp.dropna(subset=['Mean'], inplace=True)

# Hiển thị dữ liệu sau khi tiền xử lý
print("Dữ liệu sau khi tiền xử lý:")
display(df_temp.head())

# Tách dữ liệu thành GISTEMP và GCAG
df_gistemp = df_temp[df_temp['Source'] == 'GISTEMP'].copy()
df_gcag = df_temp[df_temp['Source'] == 'GCAG'].copy()
```

Bây giờ, chúng ta sẽ tạo hai bản đồ nhiệt, mỗi bản đồ cho một tiêu chuẩn (GISTEMP và GCAG), thể hiện sự ấm lên toàn cầu. Mỗi bản đồ nhiệt sẽ có các năm ở trục Y, các tháng ở trục X và màu sắc thể hiện giá trị `Mean` (sự thay đổi nhiệt độ).

```python
# === Bản đồ nhiệt cho GISTEMP ===
plt.figure(figsize=(15, 10))

# Tạo pivot table cho GISTEMP
# Sử dụng 'Year' làm chỉ mục (rows), 'Month_Name' làm cột (columns) và 'Mean' làm giá trị
heatmap_data_gistemp = df_gistemp.pivot_table(index='Year', columns='Month_Name', values='Mean')

# Đảm bảo thứ tự các tháng đúng
heatmap_data_gistemp = heatmap_data_gistemp[month_names]

sns.heatmap(heatmap_data_gistemp, cmap='coolwarm', annot=False, fmt=".1f", linewidths=.5, cbar_kws={'label': 'Nhiệt độ thay đổi (°C)'})
plt.title('Bản đồ nhiệt thay đổi nhiệt độ toàn cầu (Tiêu chuẩn GISTEMP)', fontsize=16)
plt.xlabel('Tháng', fontsize=12)
plt.ylabel('Năm', fontsize=12)
plt.tight_layout()
plt.show()

# === Bản đồ nhiệt cho GCAG ===
plt.figure(figsize=(15, 10))

# Tạo pivot table cho GCAG
heatmap_data_gcag = df_gcag.pivot_table(index='Year', columns='Month_Name', values='Mean')

# Đảm bảo thứ tự các tháng đúng
heatmap_data_gcag = heatmap_data_gcag[month_names]

sns.heatmap(heatmap_data_gcag, cmap='coolwarm', annot=False, fmt=".1f", linewidths=.5, cbar_kws={'label': 'Nhiệt độ thay đổi (°C)'})
plt.title('Bản đồ nhiệt thay đổi nhiệt độ toàn cầu (Tiêu chuẩn GCAG)', fontsize=16)
plt.xlabel('Tháng', fontsize=12)
plt.ylabel('Năm', fontsize=12)
plt.tight_layout()
plt.show()
```
